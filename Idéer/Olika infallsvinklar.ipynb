# Olika infallsvinklar

* **Övergripande översikt över problemområdet:** Leder till rekommenderade åtgärder till företag för att tackla problemet. (Genomförs med hjälp av flera intervjuer)
* **Samarbete med specifikt företag:** Längre kontakt med ett företag där vi ser över deras problematik kring Shadow IT och förmedlar anpassade åtgärder.
* **”Rollspel”:** Vi tar roller som säkerhetskonsulter och tar fram allmänna underlag för att undvika Shadow IT.
* **Dashboard:** Vi producerar en Dashboard som ett verktyg för teamledare att navigera problemen kring Shadow IT.
