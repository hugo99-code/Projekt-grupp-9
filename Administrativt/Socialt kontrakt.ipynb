# Socialt kontrakt – Grupp 9

## 1. Kommunikation & Närvaro
* **Fysisk närvaro:** Arbetet sker som standard tillsammans på plats, om gruppen inte uttryckligen kommit överens om annat.
* **Tidspassning:** Alla respekterar gruppens tid genom att komma i tid till överenskommen tid och plats.
* **Förseningar:** Vid eventuell försening ska gruppen meddelas så tidigt som möjligt med information om hur sen man beräknas bli.
* **Digital dialog:** Alla håller sig aktiva i gruppchatten inför planering av tid och plats. Nya meddelanden ska läsas så fort som möjligt och besvaras vid behov.

---

## 2. Veckoplanering
* **Måndagsplanering:** Varje måndag planeras veckans arbete och ett schema upprättas.
* **Förhinder:** Eventuellt jobb eller andra åtaganden under arbetstiderna meddelas under måndagsmötet, så att schemat kan synkas för maximal närvaro på plats.
* **Frånvarohantering:** Om någon inte kan delta beslutar gruppen gemensamt om åtgärd:
  * Personen utför sin del av arbetet på egen hand vid annat tillfälle, eller
  * Gruppen ombokar och anpassar arbetspasset till en annan tidpunkt.

---

## 3. Arbetsfördelning & Kvalitet
* **Jämn fördelning:** Arbetsinsatsen delas upp jämnt mellan alla medlemmar utifrån arbetets gång och rådande behov.
* **Eget ansvar:** Varje medlem ansvarar för att leverera sin tilldelade del i tid och med tillräcklig kvalitet.
* **Öppenhet vid hinder:** Problem eller blockeringar som uppstår ska kommuniceras till gruppen omgående.
* **Gemensamma moment:** Vid gemensamma arbetsuppgifter förväntas samtliga medlemmar bidra aktivt.

---

## 4. Beslutsfattande & Konflikthantering
* **Beslut:** Beslut fattas i första hand i samförstånd. Vid oenighet tillämpas majoritetsbeslut.
* **Takhöjd och dialog:** Gruppen upprätthåller ett öppet klimat där alla får presentera sina idéer. All feedback och kritik ska vara konstruktiv och respektfull.
* **Avtalsbrott:** Om någon bryter mot kontraktet hanteras det direkt genom en öppen diskussion inom gruppen, där personen tar ansvar för att problemet inte upprepas.
* **Eskalering:** Om problem kvarstår och inte kan lösas internt inom gruppen lyfts ärendet till kursansvarig.
