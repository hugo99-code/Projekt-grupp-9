# Definition of Done (DoD)

## Generellt (gäller alla kort på ert Scrumboard)
- [ ] Alla delmoment i kortet är genomförda och acceptanskriterierna är uppfyllda.
- [ ] Kortet är taggat med rätt kategori och kopplat till aktuell iteration och milstolpe.
- [ ] Arbetet är granskat och godkänt av kvalitetsansvarig.
- [ ] Uppdateringar och tidsåtgång är synkade inför måndagsmötet och dokumenterade i projektdagboken.

---

## DoD: Research, Intervjuer & Intressentkontakt (Hotell Tylösand)
- [ ] Intervjuanteckningar är renskrivna och strukturerade i gruppens gemensamma mapp.
- [ ] Kärninsikter och citat är taggade och överförda till underlaget för nulägesanalysen och riskkartläggningen.
- [ ] Kontaktansvarig (Anastasia) har följt upp kontakten med intervjupersonen/PO och skickat formellt tackmejl.

---

## DoD: Teoretiskt underlag & Källor
- [ ] Källan är en granskad vetenskaplig tidskrift/artikel (för att uppfylla delmålet om minst 7 vetenskapliga artiklar).
- [ ] Artikeln är sammanfattad i gruppens källmatris och kopplad till Shadow-IT eller projektstyrning.
- [ ] Referensen är fullständigt och korrekt införd i referenslistan enligt etablerat referenssystem (t.ex. APA).

---

## DoD: Rapportavsnitt & Text (Projektplan M1 & Slutrapport M2)
- [ ] Texten är inskriven direkt i det officiella huvuddokumentet (inga lösa utkast eller lokala filer).
- [ ] Språknivå och struktur möter kraven för Väl godkänt enligt kursens bedömningskriterier.
- [ ] Alla tabeller och figurer är numrerade, har informativa rubriker och refereras till i den löpande brödtexten.
- [ ] Peer review: Texten har korrekturlästs och godkänts av alla i projektet.

---

## DoD: Verksamhetsresultat & Åtgärdsförslag (Underlag till Hotell Tylösand)
- [ ] Åtgärdsförslaget adresserar en konkret identifierad risk från er Shadow-IT-kartläggning (mot målet om minst 5 åtgärdsförslag).
- [ ] Förslaget balanserar medarbetarnas behov av snabbrörlighet mot organisationens krav på informationssäkerhet och policyefterlevnad.
- [ ] Åtgärden är formulerad som en realistisk och implementerbar rekommendation som kan presenteras direkt för verksamhetsledningen.

---

## DoD: Presentationer & Kamratgranskning (M1, M2 & M3)
- [ ] Bildspel/presentationsmaterial är färdigställt, visuellt enhetligt och godkänt av alla i gruppen minst en dag före utsatt datum.
- [ ] Gruppen har genomfört en gemensam torr-repetering med tidtagning inför det muntliga examinationstillfället.
- [ ] För kamratgranskning (M3): Skriftlig återkoppling innehåller både identifierade styrkor och konstruktiva förbättringsförslag utifrån kursens betygskriterier.
