# Projektdagbok – Grupp 9

**Projektmedlemmar:** Anastasia Stojanova, Axel Bergström, Hugo Lindbärg, Oscar Nilsson

---

## Vecka 37

### 08/09 - 2026
**Uppstart och förväntningar**
Uppstart av projektet. Vi samlades för första gången för att diskutera individuella mål och förväntningar på projektet. Syftet var att hitta en gemensam riktning och upprätta ett socialt kontrakt för att se till att projektets krav är tydligt kommunicerade. Därefter fortsatte vi kort med lite diskussion kring idéer.

### 09/09 - 2026
**Idébrainstorming och problemområde**
Denna dagen hade vi ett tydligt fokus på att brainstorma idéer. Vi upptäckte att det fanns ett gemensamt intresse för cybersäkerhet och datahantering, vilket innebar att vi började undersöka projektideer inom dessa områden. För att få en bättre bild av ämnet beslutade vi oss för att genomföra intervjuer med individer inom konsultbranschen. Målet är att genom denna inblick i problemområdet förstå var problemen faktiskt finns och styra projektet efter det.

### 10/09 - 2026
**GitHub-struktur, backlog och projektöversikt**
Idag låg fokus på att komma igång med projektplanen och planera vårt arbete. Vi skapade ett repository där vi kommer samla alla dokument under projektets gång och använda oss av scrumboardet. Just det la vi en del tid på idag, fokus var på att fylla på backlogen och planera de kommande dagarnas arbete. I skrivväg kom vi igång med att skriva projektöversikten i projektplanen. 

### 11/09 - 2026
**Förberedelser, intervjubokning och metod** 
En dag där vi planerade och förberedde oss för kommande veckan. Vi bokade in 2 intervjuer med företag som vi ska ha nästa vecka och skrev intervjufrågor. Bestämde oss för att alla ska gå igenom kurslitteraturen den kommande veckan och började göra det individuellt. Skrev också ned ett socialt kontrakt för projektet.

## Vecka 38

### 14/09 - 2026
**Workshop – Projekttriangeln**
Denna dagen hade vi workshop och fick testa att genom projekttriangeln tydligt visualisera projektets ramar. Detta gjorde vi med en annan grupp och hade ni möjlighet att diskutera frågor som kvalitet, tid och resurser. Det var ett bra sätt att faktiskt prioritera olika styrparametrar samt göra målet mer konkret och tydligt. 

### 15/09 - 2026
**Avgränsning till Shadow-IT**
Fokus denna dagen låg på att smalna av undersökningsområdet. Genom diskussioner och öppen dialog med alla gruppmedlemmar landade vi i att utforska shadow- IT i konsultbranschen vidare. Dessutom kontaktade vi ytterligare företag som ex. Hotel Tylösand, för att få vidare förståelse för organisatoriska problem kopplade till området.

### 16/09 - 2026
**Planering av kommande arbete**
Vi letade källor kring området. Dessutom sammanfattade vi projektprocessen och gick vidare mot ett eventuellt samarbete med Hotel Tylösand. Vi har också fört över filer från Google Docs till Github repository och fyllt på backlogen. Till sist fortsatte vi skriva lite på projektbakgrunden.

### 17/09 - 2026
**Projektplan & bokning av intervju**
Denna dagen genomförde vi en riskanalys, förväntade resultat, intressenter, framgångsindikationer och kommunikationsplan (en mycket produktiv dag). Dessutom tog vi beslut kring ett fortsatt arbete med Hotel Tylösands organisatoriska cyber säkerhetsarbete. Vi bokade en tid för intervju med IT- ansvariga på hotellet Tony, imorgon 18/9.

### 18/09 - 2026
**Intervju med Tony**
Genomförde en intervju med Tony, som är IT- Ansvarig på Hotel Tylösand. Gav flera insikter som vi transkriberade och sammanfattade. Efter det utvecklade vi delen “problem, utmaning eller möjlighet”. Dessutom la vi upp en plan för nästa vecka och fortsatt arbete. Bland annat bokade vi intervjuer med andra anställda på Hotel Tylösand för att få lite andra perspektiv kring problemet. Under dagen bjöds också fika för att fira en produktiv vecka.
