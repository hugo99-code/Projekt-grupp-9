# Projektdagbok – Grupp 9

**Projektmedlemmar:** Anastasia Stojanova, Axel Bergström, Hugo Lindbärg, Oscar Nilsson

---

## Vecka 37

### 08/09 - 2026
* **Fokus:** Uppstart och förväntningar
* **Aktivitet:** Vi samlades för första gången för att diskutera individuella mål och förväntningar på projektet. Syftet var att hitta en gemensam riktning och upprätta ett socialt kontrakt för att se till att projektets krav är tydligt kommunicerade. Därefter fortsatte vi kort med diskussion kring idéer.

### 09/09 - 2026
* **Fokus:** Idébrainstorming och problemområde
* **Aktivitet:** Dagen ägnades åt att brainstorma idéer. Vi upptäckte ett gemensamt intresse för cybersäkerhet och datahantering, vilket ledde till att vi började undersöka projektidéer inom dessa områden. För att få en tydligare bild av ämnet beslutade vi oss för att genomföra intervjuer med personer inom konsultbranschen, i syfte att förstå var problemen faktiskt finns och styra projektet därefter.

### 10/09 - 2026
* **Fokus:** GitHub-struktur, backlog och projektöversikt
* **Aktivitet:** Fokus låg på att komma igång med projektplanen och planera arbetet. Vi skapade ett repository där vi samlar alla dokument under projektets gång och etablerade vårt Scrumboard. Mycket tid lades på att fylla på backlogen och planera de kommande dagarnas arbete. Skrivmässigt påbörjade vi projektöversikten i projektplanen.

### 11/09 - 2026
* **Fokus:** Förberedelser, intervjubokning och metod
* **Aktivitet:** Vi planerade och förberedde inför kommande vecka genom att boka in 2 intervjuer med företag och ta fram intervjufrågor. Gruppen kom överens om att alla individuellt ska läsa in sig på kurslitteraturen under den kommande veckan. Dessutom skrev vi ned och fastställde det sociala kontraktet för projektet.

---

## Vecka 38

### 14/09 - 2026
* **Fokus:** Workshop – Projekttriangeln
* **Aktivitet:** Vi deltog i en workshop och testade att visualisera projektets ramar med hjälp av projekttriangeln tillsammans med en annan grupp. Vi diskuterade avvägningar mellan kvalitet, tid och resurser, vilket gav ett bra underlag för att prioritera styrparametrar samt göra projektmålet mer konkret och tydligt.

### 15/09 - 2026
* **Fokus:** Avgränsning till Shadow-IT
* **Aktivitet:** Arbetet inriktades på att smalna av undersökningsområdet. Efter diskussioner och öppen dialog inom gruppen landade vi i att utforska **Shadow-IT inom konsultbranschen**. Vi tog även kontakt med ytterligare aktörer (bland annat Hotel Tylösand) för att få en bredare förståelse för organisatoriska utmaningar kopplade till området.
