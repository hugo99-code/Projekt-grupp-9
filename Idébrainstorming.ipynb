# Idébrainstorming

## Projektets ramar
Projektet kan exempelvis bestå i att:
* Utveckla eller utvärdera en digital lösning
* Analysera eller förbättra informationssystem i en organisation
* Bidra till ett forsknings- eller utvecklingsprojekt
* Utveckla och testa en egen digital idé eller prototyp

---

## Vad vi vet att vi inte vill
* Inte utveckla en traditionell app eller en ren hemsida

---

## Brainstormade idéer
* Anas farsas företag
* Undersöka en produkt, ett arbetssätt, ett projekt etc.
* Kombinera hård- och mjukvara
* Få barn att röra på sig mer
* Utvärdera roller inom projekt utifrån AI:s påverkan
* Råd/tips för att ligga i framkant gällande datahantering
* Utveckla ett verktyg i form av ett tillägg (Chrome Extension)
  * Automatiskt godkänna de kakor man vill ha på nya hemsidor
  * Ge en bild av vad det innebär att godkänna kakor
  * Visa hur företag tar tillvara på datan man ger ifrån sig
  * Koppla till algoritmer och cybersäkerhet

---

## Idé vi kom fram till att utforska

> **Datahantering och regelefterlevnad hos digitalt omogna företag**
> 
> En branschundersökning med fokus på exempelvis småföretagare och hantverkare:
> * **Riktlinjer:** Vad har dessa verksamheter för faktiska rutiner kring databehandling och kunduppgifter?
> * **Regelefterlevnad:** Följer de gällande lagstiftning (GDPR), eller nöjer man sig med "good enough"?
> * **Informationssystem:** Hur hanteras data i det dagliga arbetet och var uppstår riskerna?
