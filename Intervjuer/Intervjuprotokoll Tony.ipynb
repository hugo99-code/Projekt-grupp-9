* **Respondent:** Tony, IT- och systemansvarig, Hotell Tylösand
* **Intervjuare:** Projektgrupp 9
* **Fokusområde:** Cybersäkerhet, datahantering, AI-användning och Skugg-IT

---

## 1. Verksamhet, data och affärskritiska system
* **Hanterade datatyper:** Verksamheten hanterar stora volymer gästdata, inklusive kontaktuppgifter, e-postadresser och personuppgifter kopplade till vistelser, bokningar och preferenser.
* **Fragmentering och efterlevnad (GDPR):** Data är utspridd över flertalet olika system. Hanteringen försvåras av att många medarbetare (flera hundra) behöver förstå regelverket, vilket gör GDPR-utbildning resurskrävande. Gäster begär emellanåt att få sin data raderad ("rätten att bli bortglömd"), vilket är svåradministrerat i en fragmenterad miljö.
* **Fysisk säkerhetsrisk:** Känslig information skrivs i perioder ut i fysisk form på papper, vilket medför risker för obehörig åtkomst utanför de digitala systemen.
* **Driftkritiska system (24/7):** Tre operativa system utgör kärnan och tolererar inga avbrott:
  * Hotell- och rumsbokningssystemet (inkl. gästprofiler, rumsfördelning och måltidsbokningar).
  * Spasystemet (bokning och hantering av behandlingar och faciliteter).
  * Restaurang- och kassasystem (POS) (försäljning av mat och dryck).

---

## 2. Infrastruktur, IT-arkitektur och teknisk skuld
* **Hybridlösning och modernisering:** Infrastrukturen består av en hybridmiljö där lokala servrar i fastigheten synkas mot molnet via Microsoft Azure.
* **Avveckling av internetexponerade servrar:** Hotellet har fasat ut direkt internetåtkomst till egna servrar. I stället används Microsoft Entra (Application Proxy / Reverse Proxy) för att säkert tillgängliggöra interna applikationer externt, vilket markant minskat attackytan och arbetsbelastningen kring intrångsförsök.
* **Teknisk skuld (Legacy systems):** Verksamheten har funnits i över 100 år. Äldre servrar och äldre system finns kvar och är svåra att avveckla då de blockerar förändringar eller är för kostsamma att byta ut relativt nyttan så länge integrationen fungerar.
* **Leverantörsberoende:** Som medelstor aktör måste hotellet i stor utsträckning acceptera branschleverantörernas tekniska upplägg och takt. Branschen generellt uppfattas ligga efter gällande modern IT-arkitektur.
* **Incidenthistorik och ramverk:**
  * Drabbades av ett ransomware-angrepp för tio år sedan där filer började krypteras på äldre servrar; upptäcktes och isolerades inom en timme.
  * Vardagliga incidenter är få och har främst hanterats med lösenordsbyten.
  * Inget formellt ramverk (t.ex. NIST eller ISO 27001) tillämpas fullt ut idag, men kravnivåer och regleringar (såsom NIS2-tröskelvärden) bevakas löpande.

## 3. Mänskliga faktorer och säkerhetskultur
* **Utbildning och medvetenhet:** Formell, kontinuerlig cybersäkerhetsutbildning saknas. Dialogen sker främst reaktivt via mejlutskick och informella avstämningar.
* **Medarbetarbeteende:** Personalen bedöms i grunden vara relativt försiktig, men bristande kunskap och hög arbetsbelastning skapar risker för handhavandefel.

---

## 4. AI-användning, policyer och Skugg-IT
* **Verktyg i bruk:** ChatGPT och Microsoft Copilot används dagligen i organisationen. Alla anställda har tillgång till Copilot via hotellets befintliga Microsoft-licenser.
* **Saknad av AI-policy:** Det finns ingen formaliserad policy eller skriftliga riktlinjer för hur AI-verktyg får användas eller vilken data som får matas in.
* **Dataläckage och beteende:** Det förekommer att känslig data (personuppgifter eller ekonomisk information) matas in i AI-verktyg. IT-ansvarig har gett muntliga rekommendationer att undvika detta, men medarbetarna drivs av bekvämlighet och effektivitet.
* **Skugg-IT i praktiken:** IT-avdelningen har inte fullständig överblick över vilka tredjepartsappar och molntjänster medarbetarna testar på eget bevåg. Kontrollen underlättas något av en relationsnära organisation ("man känner alla"), men luckor i insynen erkänns.
* **Framtida integration:** Tredjepartsleverantörer bygger i snabb takt in AI i hotellsystemen, vilket väntas öka användningen ytterligare framöver.

## Kärninsikter för projektgruppens fortsatta arbete
* **Problemgapet bekräftat:** Det finns en tydlig diskrepans mellan medarbetarnas vilja att använda AI/SaaS för att arbeta snabbare och organisationens skydd för personuppgifter och gästdata.
* **Policy saknas – verktyg behövs:** Eftersom hotellet helt saknar en AI-policy räcker det inte med förbud. Det finns ett direkt behov av en struktur (portal/snabbgranskning) som visar medarbetarna vad som är tillåtet och förhindrar att känslig gästdata läcker till publika modeller.
* **Teknisk verklighet:** Lösningar och rekommendationer måste anpassas till en Microsoft-centrerad miljö med begränsade interna IT-resurser och ett stort beroende av externa mjukvaruleverantörer.
