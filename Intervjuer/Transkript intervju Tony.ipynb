## Sammanfattning av viktigaste insikter

* **Affärskritiska system och drift:** Verksamheten har en hybridmiljö med tre primära system igång dygnet runt: hotell- och konferensbokning, spa-system samt kassa- och restaurangsystem (POS). Vissa lokala servrar finns kvar men skyddas via Microsoft Entra Private Network Connector snarare än att exponeras öppet mot internet.
* **Historiskt arv och teknisk skuld:** Verksamheten har anor sedan över 100 år, vilket medför äldre tekniska lösningar och protokoll som inte har kunnat fasas ut helt på grund av systemberoenden.
* **Svagheter i branschen:** IT-mognaden inom besöks- och hotellbranschen ("hospitality") upplevs generellt ligga efter andra branscher. Mindre organisationer saknar dessutom förhandlingskraft mot SaaS-leverantörer, där insynen i leverantörernas säkerhetsnivå ofta är bristfällig.
* **Personuppgifter och fysiska rutiner:** Hantering av gästdata försvåras av att informationen sprids över många system samt att personalen ibland hanterar fysiska utskrifter och listor felaktigt.
* **Säkerhetskultur och utbildning:** Formell säkerhetsutbildning saknas och sker främst reaktivt via varningsmejl. Mänskliga misstag (som klick på phishing-länkar) sker, och verksamheten råkade ut för en ransomware-incident för cirka tio år sedan som dock stoppades i tid.
* **AI och skugg-IT:** Verksamheten saknar en formell policy för generativ AI. Anställda nyttjar ChatGPT och Copilot, och det förekommer att känsliga uppgifter eller siffror matas in som prompts i syfte att effektivisera det dagliga arbetet. Framtida AI-funktionalitet väntas främst tillkomma genom externa systemleverantörers uppdateringar.
