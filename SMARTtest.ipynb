# SMART-test för projektmål
**Handout – Projektkurs**

Använd SMART-testet när ni formulerar eller kvalitetssäkrar projektets mål.

> **Syfte:** Ett SMART mål gör det tydligt vad som ska uppnås, hur ni vet att ni lyckats och när resultatet ska vara klart.

---

## 1. SMART-testet

| Del | Betyder | Kontrollfrågor | Anteckning / Förbättring |
| :--- | :--- | :--- | :--- |
| **S** | **Specifikt** | • Vad ska konkret uppnås?<br>• Vem eller vilka berörs?<br>• Vad ingår och vad ingår inte? | • Utforska och underlätta kommunikationen mellan ledningen och anställda.<br>• Ledningen och anställda.<br>• **Ingår:** Användningen av shadow-IT och kommunikation.<br>• **Ingår ej:** Allt annat. |
| **M** | **Mätbart** | • Hur avgör vi om målet är nått?<br>• Vilka mått, kriterier eller bevis använder vi?<br>• Vilken nivå räknas som godkänd? | • Vi har skapat ett sätt för organisationer att undvika skadlig shadow-IT.<br>• Dagens problemsituation utifrån intervjuer och artiklar.<br>• En slutrapport som undersöker ämnet med tillräckligt många källor och en fungerande prototyp. |
| **A** | **Accepterat** | • Är målet förankrat hos projektgrupp, beställare och berörda?<br>• Förstår alla målet på samma sätt?<br>• Finns mandat att arbeta mot målet? | • Hos projektgruppen är målet förankrat.<br>• Ja.<br>• Ja. |
| **R** | **Realistiskt** | • Är målet möjligt med tid, budget, resurser och kompetens?<br>• Vilka hinder eller beroenden finns?<br>• Är ambitionsnivån rimlig? | • Ja.<br>• För brett ämne. Svårt att få kontakt med problem owners.<br>• Ja. |
| **T** | **Tidsatt** | • När ska målet vara uppnått?<br>• Finns tydliga delmål eller milstolpar?<br>• När följer vi upp? | • Den 26 oktober.<br>• Ja, varje iteration har vi uppgifter som ska göras.<br>• Efter varje iteration. |

## 2. Testa ert projektmål

### Skriv ert nuvarande projektmål:
* Hitta källor om shadow-IT för att kartlägga kommunikationsgapet mellan ledningen och anställda.
* Utveckla en teknisk prototyp (Proof of Concept) av ett verktyg som visualiserar behov och stödjer snabbare granskning av nya applikationer.
* Leverera en slutrapport som utvärderar verktygets potential att öka transparensen.

### Efter SMART-testet – skriv om målet så att det blir tydligare:
> Utforska shadow-IT, skapa ett tekniskt verktyg som hjälper transparensen mellan ledningen och anställda och skriv en slutrapport om vad problemet är och hur det kan lösas. Ska vara klar 26/10.

### Snabbbedömning

| Snabbbedömning | S | M | A | R | T |
| :--- | :---: | :---: | :---: | :---: | :---: |
| **Poäng (0–2)** | 1 | 1 | 1 | 2 | 2 |

*Tolkning: 0 = otydligt, 1 = delvis uppfyllt, 2 = tydligt uppfyllt. Ett starkt mål ligger nära 10 poäng.*
